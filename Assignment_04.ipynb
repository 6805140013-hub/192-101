## Function 2 — `calculate_fare(seat, age, bags)`

- `age` — passenger age in years (int)

2. **Business** class (rows 1–5): multiply the base fare by **2.5**.

Use `FLIGHT = 'TG105'` and start with `booked = ['1A', '12C', '20F']`. Then repeat:

## Your program

````python

def is_valid_seat(seat, booked):
    # Write your code here
    letter="ABCDEF"
    seat=seat.strip().upper()
    if not (1 < len(seat) <= 3):
        return 'invalid format'
    num_part = seat[:-1]
    seat_letter = seat[-1]
    if not num_part.isdigit():
        return 'invalid format'
    if seat_letter not in letter:
        return 'invalid letter'
    if not(1 <= int(num_part) <= 30):
        return 'row out of range'
    if seat in booked:
        return 'already booked'
    booked.append(seat)
    return 'OK'

def calculate_fare(seat, age, bags):
    # Write your code here
    base_fare=2500.0
    seat = seat.strip().upper()
    row=int(seat[:-1])
    seatNo=seat[-1]
    if 1 <= row <= 5:
        base_fare *= 2.5
    if seatNo=='A' or seatNo=='F':
        base_fare += 200
    if age < 2:
        base_fare = base_fare * 0.1
    elif age < 12:
        base_fare *= 0.75

    if bags > 1:
        base_fare += (bags-1) * 500
    return round(base_fare, 2)


def create_ticket(name, flight, seat, fare):
    # Write your code here
    seat = seat.strip().upper()
    name= name.title()
    row=int(seat[:-1])
    if 1 <= row <= 5:
        cls='Business'
    else:
        cls='Economy'
    fare= f'{fare:,.2f}'

    name_list=name.split()
    last_name=name_list[1].upper()
    if len(last_name) > 3:
        last_name=last_name[:3]
    booking_ref= f'{last_name}-{flight}-{seat}'

    result = (f'============ E-TICKET ============\n'
              f'Passenger : {name}\n'
              f'Flight    : {flight} BKK -> CNX \n'
              f'Seat      : {seat.upper()} ({cls})\n'
              f'Fare      : {fare} THB\n'
              f'Booking   : {booking_ref}')
    return(result)




# ----------------------------- MAIN PROGRAM -----------------------------
FLIGHT = 'TG105'
booked = ['1A', '12C', '20F']
counter=0
fare=0.0

while True:
    n=input("Passenger name (or done): ")
    if n.strip().upper() == "DONE":
        break
    s=input("Seat: ").strip().upper()
    valid_seat=is_valid_seat(s, booked)
    while valid_seat != 'OK':
        print(f"x {valid_seat}")
        s=input("Seat: ").strip().upper()
        valid_seat=is_valid_seat(s, booked)
    a=int(input("Age: "))
    bs=int(input("Bags: "))
    f=calculate_fare(s,a,bs)
    print(create_ticket(n,FLIGHT,s,f))
    counter+=1
    fare+=f

print('Tickets sold :',counter)
print(f'Total revenue :{fare:,.2f} THB')
print('Booked seats :',','.join(booked))
````
